- Índices de seca (Palmer Drought Severity Index)

Valor do PDSI	Interpretação
≥ +4,0	        Muito úmido
+3,0 a +3,9	    Moderadamente úmido
+2,0 a +2,9	    Levemente úmido
+1,0 a +1,9	    Condições normais (úmidas)
0,0 a +0,9	    Normal
-0,9 a 0,0	    Normal
-1,0 a -1,9	    Levemente seco
-2,0 a -2,9	    Moderadamente seco
-3,0 a -3,9	    Severamente seco
≤ -4,0	        Extremamente seco
