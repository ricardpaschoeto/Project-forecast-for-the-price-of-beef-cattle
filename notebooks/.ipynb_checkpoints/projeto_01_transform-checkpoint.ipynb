### Convertendo para Valores números valores K e M

```python
# Função para converter os valores
def converter_valor(valor):
    valor = str(valor)
    valor = valor.lower().strip()
    if 'k' in valor:
        return float(valor.replace('k', '')) * 1000
    elif 'm' in valor:
        return float(valor.replace('m', '')) * 1_000_000
    else:
        return float(valor)
```

```python
df_agro_boi_not_nan_02 = df_agro_boi_not_nan_02.copy()
df_agro_boi_not_nan_02.loc[:, 'boi_futuro_Vol.'] = df_agro_boi_not_nan_02['boi_futuro_Vol.'].apply(converter_valor)
cols_with_nan = df_agro_boi_not_nan_02.columns[df_agro_boi_not_nan_02.isnull().any()].tolist()
print(df_agro_boi_not_nan_02[cols_with_nan])
```

### Normalização de escalas

```python
import seaborn as sns
from statsmodels.tsa.seasonal import seasonal_decompose
from statsmodels.graphics.tsaplots import plot_acf, plot_pacf
from statsmodels.stats.diagnostic import normal_ad
from statsmodels.tsa.stattools import adfuller
from scipy.stats import shapiro, jarque_bera
from pmdarima import auto_arima
from sklearn.preprocessing import PowerTransformer, MinMaxScaler
```

```python
import warnings
warnings.filterwarnings("ignore", category=FutureWarning)
```

Para normalizar precisamos identificar:
- Os dados são independentes ou time series
- Time series verificamos sazonalidade
- Ajustar um modelo ARIMA/SARIMA para resíduos caso das time series
- Testar normalidade em séries temporais (resíduos de modelos)
- Aplicar transformações de acordo com a normalidade (o	Box-Cox (requer dados positivos), o	Yeo-Johnson (aceita valores negativos))

```python
# Pipeline para normalização
df_normalized = df_agro_boi_clean_encoded.copy()
dic = {}
def cols_series(df):
    df_normalized = df.copy()
    float_cols = df_normalized.select_dtypes(include=['float64']).columns.tolist()
    return float_cols

def estationarity(df, col_name):
    scaler = MinMaxScaler()
    if df[col_name].nunique() != 1:
        adf_result = adfuller(df[col_name].values)
        dic['column'] =  col_name
        dic['p-value'] =  np.round(adf_result[1], 4)
        if adf_result[1] < 0.05:
            dic['stationarity'] = True
        else:
            dic['stationarity'] = False
    else:
        df_normalized.drop(columns=[col])
        return False

    return True

def modelo_arima(serie):
    modelo_auto = auto_arima(serie, seasonal=True, m=12, stepwise=True, trace=False, error_action="ignore", stationary=dic['stationarity'])
    modelo = modelo_auto.fit(serie)
    residues = modelo.resid()

    return residues

def normality_check(serie, residues, col_name):
    _, p_valor = shapiro(residues)
    dic['Shapiro-Wilk'] = np.round(p_valor, 4)
    scaler = MinMaxScaler()
    if p_valor > 0.05:
        dic['normality'] = 'SIM'
        df_normalized[col_name] = scaler.fit_transform(serie.values.reshape(-1,1))
    else:
        pt = PowerTransformer(method='yeo-johnson')
        transformed_serie = pt.fit_transform(serie.values.reshape(-1,1))
        dic['normality'] = 'NÃO'
        df_normalized[col_name] = scaler.fit_transform(transformed_serie)
    
```

```python
float_cols = cols_series(df_normalized)
for ii, col in enumerate(float_cols):
    if estationarity(df_normalized, col):
        r = modelo_arima(df_normalized[col])
        normality_check(df_normalized[col], r, col)    
    print(dic)
```

### Criação de features temporais (lag, rolling means)

```python
import pandas as pd
import numpy as np

def add_time_features(
    df: pd.DataFrame,
    time_col: str,
    target_cols: list[str],
    lag_list=(1, 7, 28),
    roll_windows=(7, 28),
    roll_stats=("mean", "std"),
    ewm_spans=(7, 28),
    add_diffs=True,
    add_pct_change=True,
    add_calendar=True,
    freq_regularize=None,   # ex.: 'D' para diário
    min_periods=1
):
    df = df.copy()

    # Lags do targets
    for col in target_cols:
        for k in lag_list:
            df[f"{col}_lag_{k}"] = df[col].shift(k)

    # Rolling (trailing) do target
    for col in target_cols:
        for w in roll_windows:
            # por linhas
            roll = gobj[col].rolling(window=w, min_periods=min_periods)
            if "mean" in roll_stats:
                df[f"{col}_rollmean_{w}"] = roll.mean().values
            if "std" in roll_stats:
                df[f"{col}_rollstd_{w}"] = roll.std(ddof=0).values
            #if "min" in roll_stats:
             #   df[f"{col}_rollmin_{w}"] = roll.min().values
            #if "max" in roll_stats:
             #   df[f"{col}_rollmax_{w}"] = roll.max().values
            #if "median" in roll_stats:
            #    df[f"{col}_rollmed_{w}"] = roll.median().values
            #if "q10" in roll_stats:
            #    df[f"{col}_rollq10_{w}"] = roll.quantile(0.10).values
            #if "q90" in roll_stats:
             #   df[f"{col}_rollq90_{w}"] = roll.quantile(0.90).values

    # EWM (média móvel exponencial) do target
    for col in target_cols:
        for s in ewm_spans:
            ewm = df[col].apply(lambda x: x.ewm(span=s, adjust=False).mean())
            df[f"{col}_ewm_{s}"] = ewm.values

    # Diferenças e variações
     
    if add_diffs:
        for col in target_cols:
            df[f"{col}_diff_1"] = df[col].diff(1)
            df[f"{col}_diff_7"] = df[col].diff(7)
    if add_pct_change:
        for col in target_cols:
            df[f"{col}_pctchg_1"] = df[col].pct_change(1)
            df[f"{col}_pctchg_7"] = df[col].pct_change(7)

    # Lags/rollings também para preditores exógenos (ex.: preco, promo)
    # Exemplo: lag de 'preco' e rolling de 'preco'
    #for k in (1, 7):
    #    df[f"preco_lag_{k}"] = gobj["preco"].shift(k)
    #roll_p = gobj["preco"].rolling(window=7, min_periods=min_periods)
    #df["preco_rollmean_7"] = roll_p.mean().values

    # Recursos de calendário
    if add_calendar:
        dt = pd.to_datetime(df[time_col])
        df["dow"] = dt.dt.weekday              # 0=segunda
        df["is_weekend"] = (df["dow"] >= 5).astype(int)
        df["dom"] = dt.dt.day
        df["month"] = dt.dt.month
        df["weekofyear"] = dt.dt.isocalendar().week.astype(int)
        # Sazonalidade cíclica (ex.: semanal e anual)
        df["dow_sin"] = np.sin(2 * np.pi * df["dow"] / 7)
        df["dow_cos"] = np.cos(2 * np.pi * df["dow"] / 7)
        df["month_sin"] = np.sin(2 * np.pi * (df["month"] - 1) / 12)
        df["month_cos"] = np.cos(2 * np.pi * (df["month"] - 1) / 12)

        # Feriados (opcional, se biblioteca 'holidays' estiver instalada)
        #try:
        #    import holidays
        #    br_holidays = holidays.country_holidays("BR")
        #    df["is_holiday_BR"] = dt.dt.date.astype("O").map(lambda d: int(d in br_holidays))
        #except Exception:
        #    df["is_holiday_BR"] = 0  # fallback sem feriados

    return df
```
